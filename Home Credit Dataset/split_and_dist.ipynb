```python
from data_split_distribute import DataSplitDistributor
from pathlib import Path

BASE_DIR = Path.cwd()
    
# Trỏ đúng vào cấu trúc thư mục như trong ảnh của bạn
RAW_DIR = BASE_DIR / "raw"
INTERIM_DIR = BASE_DIR / "interim"

print("Đang khởi tạo tiến trình chia dữ liệu...")
    
# Khởi tạo đối tượng Distributor
distributor = DataSplitDistributor(
    raw_data_dir=RAW_DIR,
    interim_data_dir=INTERIM_DIR,
    test_size=0.2,
    random_state=42,
)

distributor.run()
    
print("Hoàn tất phân chia dữ liệu vào thư mục interim!")
```

```python
TEST_DIR = BASE_DIR / "interim" / "test"

if TEST_DIR.exists() and TEST_DIR.is_dir():
    for file_path in TEST_DIR.iterdir():
        if file_path.is_file():
            stem = file_path.stem
            suffix = file_path.suffix
            
            new_filename = f"{stem}_test{suffix}"
            new_path = file_path.with_name(new_filename)
            
            # Đổi tên file
            file_path.rename(new_path)
            print(f"Đã đổi: {file_path.name} -> {new_filename}")
            
    print("Hoàn tất đổi tên toàn bộ file trong thư mục test!")
else:
    print(f"Không tìm thấy thư mục test tại: {TEST_DIR}")
```
